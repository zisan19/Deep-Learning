## Explain the concept of backpropagation in detail. Describe how the error is propagated from the output layer to the hidden layers. Derive the mathematical expressions for weight updates using the chain rule, and explain how learning rate affects the training process.


# Write Answer 01: 
Backpropagation, which stands for “backward propagation of errors” is the fundamental learning algorithm used to train feedforward neural networks. It is a supervised learning method that calculates the gradient of the loss function with respect to all the weights in the network.

It has two main phases:<br>
1.Forward pass:Input data moves through the network layer by layer. Each layer calculates a weighted sum, applies an activation function, and produces an output. The final output is compared with the actual target using a loss function to calculate the error.<br>
2.Backward pass:The error is sent backward from the output layer to the hidden layers. Using the chain rule, the algorithm calculates how much each weight contributed to the error and updates the weights to reduce future errors.<br>

#### Error propagation from the output layer to the hidden layers:
Backpropagation sends the error backward through the network using the chain rule.

Output Layer:The error is calculated by comparing the predicted output with the actual target. This shows how wrong the network’s final answer is.

Hidden Layers:Hidden layers do not have direct target values, so their error is found from the layer after them.Each hidden neuron receives a part of the error from the next layer,depending on the weights connected to it.

So,the error of a hidden neuron is:

error from next layer × connection weight × activation derivative

# Question 03: [Marks 5]

## Discuss the vanishing gradient problems in deep neural networks. Explain why they occur during backpropagation, how activation functions contribute to this issue, and suggest at least three techniques to overcome these problems.


## Write Answer 03:
Vanishing Gradient Problem:
During backpropagation in deep networks,gradients can become extremely small as they are multiplied layer-by-layer.This causes early layers to learn very slowly or not at all.

Occurs during backpropagration:

Gradients are computed via the chain rule: product of derivatives across layers.If these derivatives are consistently < 1,the product shrinks exponentially with depth.
Role of activation functions:
Saturating functions like sigmoid or tanh have near-zero derivatives in their saturated regions.When many such layers stack,gradients vanish.

Three techniques to mitigate:

Use non-saturating activations:Derivatives are 1 for positive inputs,avoiding shrinkage.

Residual connections:Create shortcut paths for gradients to flow directly across layers.

Batch normalization:Stabilizes or normalize layer inputs, keeping activations in non-saturated regions.

## Compare Artificial Neural Networks (ANN) and Convolutional Neural Networks (CNN) in detail. Discuss their architectures, working principles, advantages, limitations, and typical use cases such as image processing, tabular data, and pattern recognition.

## Write Answer 05:
### Architecture:

ANN: 
1.Fully connected dense layers. 
2.Input is flattened to a 1D vector. 
3.Every neuron connects to all neurons in the next layer.
CNN: 
1.Stacks of convolutional, pooling, and final dense layers. 
2.Uses sliding filters or kernels to process 2D or 3D data.
3.preserving spatial structure.

### Working Principles:

ANN:
Calculates a weighted sum of all inputs,adds a bias and then applies an activation function. It learns the relationship between input and output but does not understand spatial or time based structure.

CNN:
Uses convolution to find local features then pooling to reduce size.It builds features step by step.It uses local connections and shared weights which reduces the number of parameters.

### Advantages:

ANN: Simple,flexible,fast to train on small/medium datasets,works well on structured data.

CNN:Looks at small patches of the data to find basic shapes like edges (convolution) then shrinks the data to keep only the most important parts (pooling).Because it reuses the same small filters and it saves a massive amount of memory.

### Limitations:

ANN:Demands way too much memory for large pictures.It also destroys the shape of an image by flattening it,making it easy for the model to get confused.

CNN:Terrible for regular spreadsheet data.It also needs a huge amount of training data and requires powerful computers to run.

### Typical Use Cases:

Image Processing:CNN works best for tasks like classification and detection.ANN is not suitable because it loses spatial information and needs too many parameters.

Tabular Data:ANN is better.CNN does not help unless the data is changed into a grid format.

Pattern Recognition:CNN is strong for spatial or time-based data ANN is better for general non-spatial data like financial or sensor data.

## Explain different activation functions used in neural networks, including Sigmoid, Tanh, and ReLU. Provide their mathematical formulas, advantages, disadvantages, and suitable use cases. Also discuss why ReLU is widely used in deep learning.

## Write Answer 06:
For Sigmoid:

Formula: σ(x) = 1/(1 + e⁻ˣ)

Advantages: Smooth, differentiable, outputs (0,1) → interpretable as probability

Disadvantages:Flattens at extreme inputs,triggering vanishing gradients. outputs aren’t centered around zero.relies on costly exponential operations.

Use Case:Binary classification output layer,probabilistic heads.

for Tanh:

Formula: tanh(x) = (eˣ − e⁻ˣ) / (eˣ + e⁻ˣ)

Advantages: Zero-centered (-1,1). stronger gradients near 0 than sigmoid.

Disadvantages: Still saturates at large magnitudes, causing gradient decay. similarly expensive to compute due to exponentials.

Use Case: Hidden layers in RNNs, legacy networks, tasks benefiting from zero-centered activations

For ReLU

Formula: f(x) = max(0, x)

Advantages: Extremely fast to compute,non-saturating for x>0,promotes sparse activations,accelerates convergence

Disadvantages:Neurons may stop learning,Output is not centered Suitable Use cases.
Use Case: Default hidden-layer activation in CNNs, MLPs, and most modern deep architectures.

Discuss of why ReLU is widely used in deep learning:

ReLU dominates deep learning because it preserves gradients for positive inputs, runs fast through simple thresholding, encourages sparse activations for better generalization, consistently delivers superior training speed and accuracy, and easily extends into variants like Leaky ReLU, ELU, and GELU to handle edge cases.

# Question 08: [ Marks 10 ]

## Explain the concept of pooling in CNNs. Compare MaxPooling and AveragePooling in terms of operation, advantages, and impact on feature extraction. Discuss how pooling helps in reducing overfitting and computational complexity.

Write Answer 08:

Pooling Concept:

A downsampling operation that slides a fixed window over feature maps and aggregates values, shrinking spatial dimensions while preserving key patterns and introducing translation invariance.

MaxPooling vs AveragePooling:
Operation:MaxPooling selects the highest value per window but AveragePooling computes the mean.

Advantages:MaxPooling retains strong activations and is robust to noise but AveragePooling smooths outputs, preserving broader context and reducing activation variance.  
Feature Extraction Impact:MaxPooling emphasizes dominant, localized features but AveragePooling captures distributed, global patterns.

Reducing Overfitting & Computational Complexity:

Complexity:
Reduces the size of feature maps, which greatly lowers the number of computations, memory usage, and overall processing cost in later layers.

Overfitting Control:
Removes exact location details of features, making the model less sensitive to small changes. This helps the network learn more general and stable patterns instead of memorizing noise in the training data.

```python
import os
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import transforms
from torch.utils.data import Dataset,DataLoader,Subset
from PIL import Image
import kagglehub
```

```python
path = kagglehub.dataset_download("mohitsingh1804/plantvillage")
print("path",path)
```

```python
print(os.listdir(path))
```

```python
torch.manual_seed(42)
device = torch.device('cuda'if torch.cuda.is_available() else 'cpu')
print(f" Using Device {device}")
```

```python
TRAIN_PATH = os.path.join(path, "PlantVillage","train")
VAL_PATH = os.path.join(path, "PlantVillage","val")
```

```python
transform = transforms.Compose(
    [
        transforms.Resize( (128,128) ),
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.5]*3, std=[0.5]*3)
    ]
)
```

```python
transform = transforms.Compose(
    [
        transforms.Resize( (128,128) ),
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.5]*3, std=[0.5]*3)
    ]
)
```

```python
from genericpath import isfile

class MultiClassClassfication(Dataset):
  def __init__(self, root_dir,transform=None) :
    super().__init__()

    self.samples = []
    self.transform = transform
    self.classes = sorted([d for d in os.listdir(root_dir) if os.path.isdir(os.path.join(root_dir,d))])
    self.class_to_idx= {cls_name : idx  for idx,cls_name in enumerate(self.classes)}
    for class_name in self.classes:
      class_path = os.path.join(root_dir,class_name)
      for img_name in os.listdir(class_path):
        img_path = os.path.join(class_path,img_name)
        if os.path.isfile(img_path):
          label = self.class_to_idx[class_name]
          self.samples.append( (img_path,label) )
  def __len__(self):
    return len(self.samples)
  def __getitem__(self, idx) :
    img_path, label = self.samples[idx]
    image = Image.open(img_path).convert("RGB")
    if self.transform:
      image = self.transform(image)
    return image, label
```

```python
train_dataset_full = MultiClassClassfication(TRAIN_PATH, transform)
test_dataset_full = MultiClassClassfication(VAL_PATH, transform)
num_classes = len(train_dataset_full.classes)

print("Number of classes:", num_classes)
print("Full train size:", len(train_dataset_full))
print("Full test size:", len(test_dataset_full))
```

```python
train_dataset = Subset(train_dataset_full, list(range(min(100, len(train_dataset_full)))))
test_dataset = Subset(test_dataset_full, list(range(min(100, len(test_dataset_full)))))
```

```python
pin = True if device.type == 'cuda' else False
train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True, pin_memory=pin)
test_loader = DataLoader(test_dataset, batch_size=32, shuffle=False, pin_memory=pin)
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class MyCNN(nn.Module):
    def __init__(self, input_channels=3, num_classes=10):
        super().__init__()

        # for block 1
        self.block1 = nn.Sequential(
            nn.Conv2d(3, 32, kernel_size=3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(inplace=True),
            nn.Conv2d(32, 32, kernel_size=3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2) 
        )

        # for block 2
        self.block2 = nn.Sequential(
            nn.Conv2d(32,64, 3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True),
            nn.Conv2d(64,64, 3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2)
        )

        # for block 3
        self.block3 = nn.Sequential(
            nn.Conv2d(64,128, 3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(inplace=True),
            nn.Conv2d(128,128, 3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2)
        )

        # for block 4
        self.block4 = nn.Sequential(
            nn.Conv2d(128,256, 3, padding=1),
            nn.BatchNorm2d(256),
            nn.ReLU(inplace=True),
            nn.Conv2d(256,256, 3, padding=1),
            nn.BatchNorm2d(256),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2)
        )

        # Classifier
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(256*8*8,512),
            nn.ReLU(),
            nn.Dropout(0.4),
            nn.Linear(512, 128),
            nn.ReLU(),
            nn.Dropout(0.4),
            nn.Linear(128, num_classes)
        )
    def forward(self, x):
        x = self.block1(x)
        x = self.block2(x)
        x = self.block3(x)
        x = self.block4(x)
        x = self.classifier(x)
        return x
```

```python
model = MyCNN(num_classes=num_classes).to(device)
```

```python
learning_rate = 0.001
epochs = 10
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=learning_rate)
```

```python
for epoch in range(epochs):
    model.train()
    total_loss = 0
    for batch_features, batch_labels in train_loader:
        batch_features = batch_features.to(device)
        batch_labels = batch_labels.to(device)
        outputs = model(batch_features)
        loss = criterion(outputs, batch_labels)
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        total_loss += loss.item()
    avg_loss = total_loss / len(train_loader)
    print(f"Epoch {epoch+1}/{epochs}, Loss: {avg_loss:.4f}")
```

```python
def evaluate(loader):
    model.eval()
    total, correct = 0, 0
    with torch.no_grad():
        for batch_features, batch_labels in loader:
            batch_features, batch_labels = batch_features.to(device), batch_labels.to(device)
            outputs = model(batch_features)
            _, predicted = torch.max(outputs, 1)
            total += batch_labels.size(0)
            correct += (predicted == batch_labels).sum().item()
    return correct / total
```

```python
test_acc = evaluate(test_loader)
print("Test Accuracy:", test_acc)
```

Here is four convolution blocks.Each block learns deeper and more complex features.Filters increase from 32 → 64 → 128 → 256, so the network captures more detailed patterns at deeper levels.

The convolution blocks extract features,the fully connected layers classify them and dropout + batch normalization help improve generalization and stable training.
