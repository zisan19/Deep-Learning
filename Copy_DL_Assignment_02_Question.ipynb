# Question 01: [ Marks 10 ]
Explain gradient descent in detail. What is the main idea behind it? How does it help in minimizing the loss function? Explain the role of learning rate and describe what happens if the learning rate is too high or too low.

**Write** Answer 01:Gradient descent is an optimization method that helps find the parameters that minimize a loss function.It is foundational in machine learning and deep learning because many real-world models cannot be solved analytically,and gradient-based methods offer a practical way to find good minima of a complicated,high-dimensional function.
Main idea behiend of the gradient of a function points in the direction of steepest increase of that function with respect to it's parameters.if we want to minimize the loss, we should move in the opposite direction of the gradient.the  formula of gradient descent,
b(new)=b(old)-η.∇J

How gradient descent minimizes loss function:
*Start with an initial guess for the model parameters (weights and biases), often random.

*Compute the gradient of the loss function with respect to each parameter. The gradient is a vector of partial derivatives that points in the direction of the steepest ascent. Since we want to minimize the loss, we move in the opposite direction (negative gradient).

*Update the parameters using the formula:b(new)=b(old)-η.∇J(b)
where,b=parameters(weights)
η = learning rate
∇J(b)=gradient of loss function J with respect to parameters b

*Repeat steps 2-3 until convergence.

the role of learning rate:The learning rate (η) is step size in gradient descent. It controls how far it move with each update.
Too high (η large):The model can overshoot or bounce around the loss surface, causing oscillations or even divergence.
Too low (η small):Updates are tiny, so training is very slow progress may feel like stagnation, especially on flat regions, taking far more time to reach a good solution.

What is forward propagation in an artificial neural network? Explain step by step how data moves from the input layer to the output layer.

## Write Answer 02:
Forward propagation is the process where a neural network transforms input data into predictions or outputs.It computes the networks activation values from inputs to hidden layers to output,using the current weights and biases at each layer.
Now we see the step by step how data moves from the input layer to the output layer:

step1.Input layer receives data:
The network starts with the input vector.Each neuron in the input layer corresponds to one feature.

step2.Weights and Biases:Data gets on the "weight train" - multiplied by connection weights.At each hidden neuron,Sum up all incoming weighted data + add a "bias ticket".Pass through an activation function (like ReLU): "Should this neuron fire?".Result becomes the hidden layer's output.

step3.Transfer Between Hidden Layers:

*Repeat Step 2's process:

*Multiply by weights

*Sum + bias

*Apply activation function

*Data gets transformed at each stop

step4.Arrive at Output Terminal:Last hidden layer's output → multiplied by final weights → added to output bias then Final activation : "Is this a cat (90%), dog (7%), or bird (3%)?" and  the last Network makes its prediction


# Question 03: [ Marks 15 ]
Why do we need activation functions in neural networks? What would happen if we only used linear transformations without any activation function?

## Write Answer 03:
Activation function are essential in neural networks because Without activation functions,no matter how deep your neural network is,it's just fancy linear regression and that's weak sauce for real world problems.

if we only used linear transformations without any activation function then every layer would compute z(l)=W(l)a(l-1)+b(l) and a(l)=z(l) Since there is no activation function, every layer performs only a linear transformation. When multiple such layers are stacked, the entire network simplifies into a single linear transformation:

y_hat=W(effective)x+b(effective)
That means:

No added modeling power: You could replace the whole multi-layer network with a single linear model,since composing linear functions just gives another linear function.

Limited performance: Many real problems have strong non-linear patterns, so a purely linear network can't learn beyond basic linearly separable data and will underperform compared with even modestly-layered non-linear models.

No advantage to depth: Adding layers doesn't help—depth without non-linearity doesn't create new representational abilities, so you lose the main reason for using neural networks over simpler linear methods.

# Question 04: [ Marks 15 ]

Explain the sigmoid activation function. Where is it used? What are its advantages and disadvantages? What problem does it cause during training?

## Write Answer 04:
The sigmoid activation function is a non-linear function that maps real-valued inputs to values in the range (0,1).So,easily we called It's an S-shaped curve that squeezes any number into a range between 0 and 1.It's defined as:

σ(x)=1/(1+e^(-x))

it's used Output layer for binary classification and Hidden layers.For hidden layers,In early neural-network design,sigmoid was used inside hidden layers too but today it's less common there compared with ReLU due to training challenges.

Advantages:
*Smooth, differentiable - Essential for gradient descent

*Clear interpretation - Outputs can be read as probabilities

*Bounded output (0 to 1) - Never explodes to huge numbers

*Monotonic - Always increasing, easy to analyze

Disadvantages:Vanishing gradients,Not zero-centered,Computational cost and Sensitivity in some ranges.

The main training problem caused by the sigmoid activation function is vanishing gradients. Because sigmoid maps inputs z to the range (0,1)and its curve becomes flat when z is very large positive or very negative, its derivative σ'(z)=σ(z)(1-σ(z)) becomes very small near the extremes.

# Question 05: [ Marks 10 ]
Explain the softmax activation function in simple words. Why is it used in multi-class classification problems? How is it different from sigmoid?

## Write Answer 05:
Softmax is an activation function that takes a set of raw numbers from the output layer and converts them into probabilities values between 0 and 1 that sum to 1 for a set of classes.

Softmax is used in multi-class classification because it works naturally with the cross-entropy loss function. It converts the model's outputs into a probability distribution, where all class probabilities sum to 1.

Softmax vs. Sigmoid: The Key Difference

Softmax--------vs--------Sigmoid

Outputs are competitive------------Outputs are independent

Probabilities sum to 1---------------Each output is 0-1 independently

Used when classes are exclusive------------Used when classes can co-exist

Multi-class: Cat vs. Dog vs. Bird------------Multi-label: Contains [cat,dog]in same image

"Choose one winner"---------------"Yes/no for each category separately"


# Question 06: [ Marks 20 ]

Explain the structure of an Artificial Neural Network in detail. Describe the role of the input layer, hidden layers, and output layer. How do weights, biases, and activation functions work together to transform input data into predictions? Illustrate your explanation with a simple example.

## Write Answer 06:
An Artificial Neural Network is a computational model inspired by the structure and functioning of the human brain. It consists of a network of interconnected nodes, called artificial neurons, that process input information and generate an output.

The basic structure has three main parts: input layer,hidden layers and output layer.All working together with weights, biases, and activation functions.

*Input Layer: This is the first layer of the networks starting point.It receives the raw data that describe the example you want the model to evaluate.

*Hidden Layers: These are the layers between the input and output.They do most of the calculations and learning.

*Output Layer: This is the final layer that produces the prediction.The shape of the output depends on the task: a single neuron can predict a number or several neurons can predict probabilities for classes.
